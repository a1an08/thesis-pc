# Train YOLOv11 for Printer Defect Detection
This notebook trains a YOLOv11 nano model on a custom dataset defined in `data.yaml`.

```python
from ultralytics import YOLO

# 1. Load a pretrained YOLOv11 model (n = nano, s = small, m = medium)
# We use v11n for speed on the Raspberry Pi
model = YOLO("yolo11n.pt") 
```

```python
# 2. Train the model
results = model.train(
    data="data.yaml", 
    epochs=100, 
    imgsz=640, 
    batch=16, 
    device=0, # Use device=0 for GPU, device='cpu' for CPU
    project="printer_defect_detection",
    name="yolo11_baseline"
)
```

```python
# 3. Validate the model
metrics = model.val()
```
